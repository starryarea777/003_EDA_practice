```python
arrival = pd.to_datetime(df['입항일시'], errors= 'coerce') #입항시간

departure = pd.to_datetime(df['출항일시'], errors= 'coerce') #출항시간

stay = (departure - arrival).dt.total_seconds() / 3600 #날짜를 초로 바꾸고 다시 시간으로 바꿈
stay #계류시간
```

```python
valid = arrival.dt.year.eq(df['입항일시']) & stay.between(0, 24 * 30) #30일동안 머문 선박이 있는가 체크

stay[valid]
```

```python
date_cols = [ '입항일시', '출항일시', '출항예정일시' ]

for column in date_cols:
    df[column] = pd.to_datetime(df[column], errors='coerce')

df['체류시간'] = (df['출항일시'] - df['입항일시']).dt.total_seconds() / 3600

df['출항지연시간'] = (df['출항일시'] - df['출항예정일시']).dt.total_seconds() / 3600

df[['체류시간', '출항지연시간']]

valid_stay = (
    df['입항일시'].dt.year.eq(df['입항연도'])
    & df['체류시간'].between(0, 24 * 30)
)

valid_schedule = (
    valid_stay
    & df['출항예정일시'].between(
        df['입항일시'] - pd.Timedelta(days=1),
        df['입항일시'] - pd.Timedelta(days=30),
    )
)

valid_stay.sum()
```

```python
pd.Series({
    '원본 행 수': len(df),
    '입항연도와 입항일시 연도 불일치': df['입항일시'].dt.year.ne(df['입항연도']).sum(),
    '음수 체류시간': df['체류시간'].lt(0).sum(),
    '30일 초과 체류시간': df['체류시간'].gt(24 * 30).sum(),
    '유효 체류시간 비율': valid_stay.mean() * 100,
    '완전중복행': df.duplicated().sum()
})
```

```python
stay_qualtile = df.loc[valid_stay, '체류시간'].quantile([0.5, 0.75, 0.9, 0.95, 0.99])
stay_qualtile.index = [ '중앙값', '0.75', '0.90', '0.95', '0.99']
stay_qualtile
```
