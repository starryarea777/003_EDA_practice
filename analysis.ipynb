```python
pd.DataFrame({
    '자료형': df.dtypes.astype('str'),
    '비결측 수': df.notna().sum(),
    '결측 수': df.isna().sum(),
    '결측률(%)' : df.isna().mean() * 100,
    '고유값 수': df.nunique(dropna=True)
})
```

```python
df.describe().T
```

```python
terminal_col = []

for column_name in df.columns:
    if column_name not in [ '년도', '합계' ]:
        terminal_col.append(column_name)
        
print(terminal_col)
```

```python
new_port = []

for column_name in df.columns:
    if column_name.startswith('신항'):
        new_port.append(column_name)
        
print(new_port)
```

```python
df_1 = df.copy()

df_1['부두합계'] = df_1[terminal_col].sum(axis=1)
```

```python
df_1['신항합계'] = df_1[new_port].sum(axis=1)
```

```python
df_1['신항비중'] = (df_1['신항합계'] / df_1['부두합계']) * 100
```

```python
df_1[['년도', '부두합계', '신항합계', '신항비중']]
```
