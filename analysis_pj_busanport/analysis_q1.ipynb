```python
import pandas as pd
import numpy as np
```

'연월' 기준으로 물동량이 기록된 '부산항만공사_물동량 예측 데이터'로 계절성을 파악한다.

```python
# 데이터파일 불러오기

p_CSV_PATH = ('C:/Users/user/Desktop/data_practice/datalist/prediction.csv')

df_p = pd.read_csv(p_CSV_PATH, encoding='cp949')

df_p
```

```python
# '연월' 기준 내림차순 정렬 후, '수출입구분'으로 나누어진 물동량을 하나의 '물동량' 수치로 그룹화한다.
df_p_sorted = df_p.sort_values(by='연월', ascending=True).reset_index(drop=True)

df_p_sorted

df_p_sorted_yearlymerged = df_p_sorted.groupby('연월')['물동량'].sum()

df_p_sorted_yearlymerged

# '연월' 기준으로 하는 것 보다 '월' 기준으로 합쳐서 '물동량' 수치로 그룹화하는 것이 낫다고 판단.
```

```python
# '월' 기준으로 합쳐서 '물동량' 수치로 그룹화
df_p_monthlymerged = df_p.groupby('월',as_index=False)['물동량'].sum()

df_p_monthlymerged
```

```python
# 위 df을 통해 전체 물동량의 평균을 구한다.
mean_cargo = df_p_monthlymerged['물동량'].mean()
print(f'전체 평균 물동량: {mean_cargo:,.2f}')
```

```python
# 월별 물동량에서 전체 평균 물동량을 뺀다.

df_p_monthly = df_p_monthlymerged['물동량'] - mean_cargo
df_p_monthly
```
